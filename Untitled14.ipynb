```python
import streamlit as st
import numpy as np
import matplotlib.pyplot as plt

st.title("Tax Calculator Web App")
st.write("Comparison: Old Regime vs New Regime vs Flat Tax (15%)")
st.write("Includes 4% Health & Education Cess")

# ----------------------
# OLD REGIME
# ----------------------
def old_regime(income):
    tax = 0

    if income <= 250000:
        tax = 0
    elif income <= 500000:
        tax = (income - 250000) * 0.05
    elif income <= 1000000:
        tax = 12500 + (income - 500000) * 0.20
    else:
        tax = 112500 + (income - 1000000) * 0.30

    cess = tax * 0.04
    return tax + cess


# ----------------------
# NEW REGIME
# ----------------------
def new_regime(income):
    tax = 0

    if income <= 300000:
        tax = 0
    elif income <= 600000:
        tax = (income - 300000) * 0.05
    elif income <= 900000:
        tax = 15000 + (income - 600000) * 0.10
    elif income <= 1200000:
        tax = 45000 + (income - 900000) * 0.15
    elif income <= 1500000:
        tax = 90000 + (income - 1200000) * 0.20
    else:
        tax = 150000 + (income - 1500000) * 0.30

    cess = tax * 0.04
    return tax + cess


# ----------------------
# FLAT TAX
# ----------------------
def flat_tax(income):
    tax = income * 0.15
    cess = tax * 0.04
    return tax + cess


# ----------------------
# USER INPUT
# ----------------------

income = st.number_input("Enter Annual Income (₹)", min_value=0, value=500000)

if st.button("Calculate Tax"):

    old_tax = old_regime(income)
    new_tax = new_regime(income)
    flat = flat_tax(income)

    st.subheader("Tax Summary")

    st.write("Old Regime Tax: ₹", round(old_tax, 2))
    st.write("New Regime Tax: ₹", round(new_tax, 2))
    st.write("Flat Tax (15%): ₹", round(flat, 2))

    lowest = min(old_tax, new_tax, flat)

    if lowest == old_tax:
        st.success("Old Regime gives lowest tax")
    elif lowest == new_tax:
        st.success("New Regime gives lowest tax")
    else:
        st.success("Flat Tax gives lowest tax")

    # Graph 1
    incomes = np.linspace(0, 2000000, 100)

    old_list = [old_regime(i) for i in incomes]
    new_list = [new_regime(i) for i in incomes]
    flat_list = [flat_tax(i) for i in incomes]

    plt.figure()
    plt.plot(incomes, old_list, label="Old Regime")
    plt.plot(incomes, new_list, label="New Regime")
    plt.plot(incomes, flat_list, label="Flat Tax")
    plt.xlabel("Income")
    plt.ylabel("Total Tax")
    plt.title("Income vs Total Tax")
    plt.legend()
    st.pyplot(plt)

    # Graph 2
    plt.figure()
    eff_old = [(old_regime(i)/i)*100 if i>0 else 0 for i in incomes]
    eff_new = [(new_regime(i)/i)*100 if i>0 else 0 for i in incomes]
    eff_flat = [(flat_tax(i)/i)*100 if i>0 else 0 for i in incomes]

    plt.plot(incomes, eff_old, label="Old Regime")
    plt.plot(incomes, eff_new, label="New Regime")
    plt.plot(incomes, eff_flat, label="Flat Tax")
    plt.xlabel("Income")
    plt.ylabel("Effective Tax Rate (%)")
    plt.title("Income vs Effective Tax Rate")
    plt.legend()
    st.pyplot(plt)
```
